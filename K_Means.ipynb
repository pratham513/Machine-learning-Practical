```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
X,y = make_blobs(n_samples=300,centers=4,random_state=42)
```

```python
KMeans_labels = KMeans(n_clusters=4,random_state=42).fit_predict(X)
```

```python
kmeans_model = KMeans(n_clusters=4, random_state=42)
kmeans_model.fit(X)
KMeans_labels = kmeans_model.labels_
centroids = kmeans_model.cluster_centers_
```

```python
plt.scatter(centroids[:,0],centroids[:,1],s=100,marker='x', label='Centroids')
plt.scatter(X[:,0],X[:,1],c=KMeans_labels,cmap='viridis',marker='s',edgecolors='k', alpha=0.5,label='k-mean Predicted')
```

```python
plt.scatter(X[:,0],X[:,1],c=KMeans_labels,cmap='viridis',marker='s',edgecolors='k', alpha=0.5,label='k-mean Predicited')
```

```python
plt.scatter(centroids[:,0],centroids[:,1],s=100,marker='x', label='Centroids')
plt.legend()
plt.show()
```
